## Fact questions

**1. How are AI, ML and DL related?**

AI (Artificial Intelligence), ML (Machine Learning) and DL (Deep Learning) relate to each other as three nested sets, where each concept is a subset of the previous one.

AI is the broadest concept and refers to systems that can perform tasks that normally require human intelligence, such as reasoning, planning, recognizing patterns, or making decisions. AI can be built in several ways, e.g. with hand-written rules (expert systems) or with learned models.

ML is a subset of AI where the system is not programmed with explicit rules for every scenario, but instead learns patterns and relationships directly from data. An ML model is trained on examples and then generalizes to new, unseen observations.

DL, in turn, is a subset of ML based on artificial neural networks with many layers ("deep" networks). DL is particularly powerful for unstructured data such as images, audio and text, since the network itself learns relevant representations (features) from the raw data.

Summary: **AI -> ML -> DL**.
