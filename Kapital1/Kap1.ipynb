## Fact questions

**1. How are AI, ML and DL related?**

AI (Artificial Intelligence), ML (Machine Learning) and DL (Deep Learning) relate to each other as three nested sets, where each concept is a subset of the previous one.

Artificial intelligence is a broad field, which refers to the use of technologies to build machines and computers that have the ability to mimic cognitive functions associated with human intelligence, such as being able to see, understand, and respond to spoken or written language, analyze data, make recommendations, and more. 

From a capability standpoint, AI is broadly categorized into three types:

Artificial Narrow Intelligence (ANI), which is the specialized AI common today (like image recognition)
Artificial General Intelligence (AGI), which is human-level AI capable of any intellectual task
Artificial Super Intelligence (ASI), which is theoretical AI that surpasses human intellect. While ANI is prevalent, the pursuit of AGI is a major focus of current research.

**ML** Machine learning is a subset of artificial intelligence that automatically enables a machine or system to learn and improve from experience. Instead of explicit programming, machine learning uses algorithms to analyze large amounts of data, learn from the insights, and then make informed decisions. 

Machine learning algorithms improve performance over time as they are trained—exposed to more data. Machine learning models are the output, or what the program learns from running an algorithm on training data. The more data used, the better the model will get.

**DL**, in turn, is a subset of ML based on artificial neural networks with many layers ("deep" networks). DL is particularly powerful for unstructured data such as images, audio and text, since the network itself learns relevant representations (features) from the raw data.

Summary: **AI -> ML -> DL**.

**2. What are the four problem categories within ML?**

The book (Section 1.2, p. 16) defines four problem categories within machine learning: **regression**, **classification**, **dimensionality reduction** and **clustering**.

- **Regression**: the output (y) takes continuous values, i.e. it can take any value within a range. Example from the book: predicting a person's income based on age — income can be 47,831 kr or 32,421.5 kr, any value in between is possible.
- **Classification**: the output (y) takes discrete values. If there are exactly two possible values it is called *binary classification* (book example: predicting whether a person is a man or a woman); if there are more than two possible values it is called *multiclass classification* (book example: predicting whether a vehicle is a bicycle, car, bus or "other" — four classes).
- **Dimensionality reduction**: input data (x) is simplified by transforming it into a space with lower dimensionality, for example using a model called Principal Component Analysis (PCA). A common reason to reduce dimensionality is to make the training of regression or classification models faster.
- **Clustering**: only input data (x) is available, and the goal is to divide it into different groups/clusters. The book's example is customer segmentation, where customers are grouped based on preferences and purchasing power so that companies can tailor products and marketing to each segment.

The book also notes a broader naming convention: regression and classification are jointly called **supervised learning** (vägledd lärande), because y "guides" x during training; clustering and dimensionality reduction are jointly called **unsupervised learning** (icke-vägledd lärande), because only the input data x is used and no y guides the training.

*Source: Section 1.2, pp. 16, 19.*

**3. Explain the following:**

**a) The purpose of splitting data into training, validation and test data**

**Training data**: Used directly to fit and train different machine learning models. 

**Validation data**: Used to evaluate candidate models trained on the training data, compare their performance via evaluation metrics, and select the best model. This ensures that models are not selected solely because they memorized the training set (overfitting).

**Test data**: Kept completely untouched throughout the modeling process and evaluated only once at the very end. Its purpose is to obtain an unbiased estimate of the final model's generalization ability on new, unseen data, guarding against the risk that a model performed well on the validation set purely by random chance. 

**b) K-fold cross-validation**

The training data is split into k equally sized parts ("folds"). The model is trained k times, where one fold is used for validation and the remaining k-1 folds for training. The k performance scores are averaged, which gives a more robust estimate than a single train/validation split and makes use of all the data for both training and validation.
The resulting evaluation metrics across all $k$ iterations are averaged to produce a more reliable estimate of model performance. It reduces the risk of poor model selection caused by a single, unrepresentative validation split, though it requires training the model $k$ times

**c) Root mean squared error (RMSE)**

$$\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^n (y_i - \hat{y}_i)^2}$$
1. First the difference between the true value yᵢ and the predicted value ŷᵢ is computed: (yᵢ − ŷᵢ). This is called the **Error**.
2. Because these differences can be positive or negative, they can cancel out when summed. The book's own example: if y₁=10, y₂=8, ŷ₁=12, ŷ₂=6, then (y₁−ŷ₁)+(y₂−ŷ₂) = (10−12)+(8−6) = (−2)+(2) = 0, even though neither prediction was correct. So the differences are squared instead: (yᵢ−ŷᵢ)². This is called the **Squared Error**.
3. The mean of the squared errors is computed: (1/n)Σ(yᵢ−ŷᵢ)². This is called the **Mean Squared Error (MSE)**.
4. Because squaring changes the unit (e.g. predicting a salary in kr gives squared error in kr², hard to interpret), the square root is taken to return to the original unit — the **Root Mean Squared Error**.

The book notes that RMSE can loosely ("a bit sloppily") be interpreted as the average distance between the observed values (y) and the predicted values (ŷ). It also notes that since RMSE and MSE differ only by a square root (a monotonic transformation, so 100 > 9 ⇔ √100 > √9 ⇔ 10 > 3), they always rank models identically; MSE is often used for ranking models instead of RMSE because taking the square root is an extra operation that can take longer to compute.



**d) Hyperparameter vs. parameter**

The book defines these on p. 21 (and demonstrates them in Section 1.3.4, pp. 35–41): **hyperparameters** control how a model learns, and answer the question "how do we learn"; **parameters** are things a model has learned, and answer the question "what have we learned". Grid search (Section 1.3.5) is used to optimize hyperparameters.

The book's own worked example uses linear regression's `fit_intercept` hyperparameter, which can be set to `True` or `False`, and is chosen automatically via `GridSearchCV` over the candidate values `{'fit_intercept': [True, False]}`. After running the grid search, the book prints the learned parameters of the resulting best model: "Best Model Coefficients. Intercept: 0.0, Slope: 0.8" — here the intercept and slope are the *parameters* the model has learned, while `fit_intercept` is the *hyperparameter* that was chosen before/during training to control how the model was allowed to learn.

```python
from sklearn.model_selection import GridSearchCV

model = LinearRegression()
hyperparameters = {'fit_intercept': [True, False]}
grid_search = GridSearchCV(estimator=model, param_grid=hyperparameters,
    scoring='neg_mean_squared_error', cv=2)
grid_search.fit(example_data[["x"]], example_data["y"])

****On refit=True****: reading the official scikit-learn documentation for `GridSearchCV` (as the book does, p. 41), the hyperparameter `refit` has the default value `True`, and the documentation states: *"Refit an estimator using the best found parameters on the whole dataset."* The book explains that once the grid search has been performed (`.fit()`) and the optimal hyperparameters have been found, the model is retrained (refit) using those best hyperparameters on the whole dataset that was passed in, and this refitted, optimal model is what is saved — for example accessible as `grid_search.best_estimator_` — so it can be used directly afterwards to make predictions, as the book demonstrates with `grid_search.predict([[6]])

**f) Categorical data**

**Nominal data** has no inherent order (e.g. `origin`: europe/japan/usa). **Ordinal data** has a natural ranking (e.g. education level). Since ML models require numeric input, categorical data is re-encoded: **one-hot encoding** creates a binary column per category (k columns), **dummy-variable encoding** does the same but omits one category (k-1 columns, `drop_first=True`) to avoid multicollinearity, and **ordinal encoding** translates categories into integers that preserve the order — only suitable for ordinal data.

**g) Feature engineering**

The process of selecting which variables to use (variable/feature selection), deriving and creating new variables using domain knowledge (e.g., calculating average rooms per household or binary indicators), and transforming existing features (e.g., standardizing data using StandardScaler to have a mean of 0 and standard deviation of 1).

**h) Principle of parsimony**

Given two models with similar performance, the simplest one should be preferred (cf. Occam's razor) — easier to interpret, often better generalization, and cheaper to run. Complexity should only be accepted if it provides a clear benefit.

**4. "A model is a simplification of reality" – what does this mean?**

A model can never capture all the details, noise and complexity of reality. It is built on a selection of variables and assumptions about how they relate to what one wants to predict, and is therefore always an approximation. This is a strength rather than a weakness — a model that tried to reproduce reality in all its detail would become unreasonably complex and would overfit to exactly the data it has seen.

**5. What does it mean for a model to be "overfitted"?**

The model has learned the training data so well — including its noise and random variations  that it performs very well on the training data but considerably worse on new, unseen data. It has memorized details instead of learning general, underlying patterns.

**6. "Higher is better" in scikit-learn scoring — what does this mean?**

Scikit-learn's convention is that a higher scoring value should always mean a better model, regardless of whether the underlying metric is actually "good when low" (like RMSE/MSE). See question 9 for how this is solved in practice (`neg_mean_squared_error`).

**7. What are cross-sectional data, time series data and panel data?**

- **Cross-sectional data**: several different units at one and the same point in time, e.g. the income and education of 1000 people during a given month.
- **Time series data**: the same unit over several points in time, e.g. a stock's daily closing price over a year.
- **Panel data**: several units over several points in time, e.g. unemployment in Sweden's 21 regions measured every quarter over ten years.

## Reasoning questions

**8. Examples of real-world applications within ML**

Natural language processing (NLP) & chatbots: Creating automated conversational assistants such as ChatGPT, performing sentiment analysis, or auto-generating text.

Computer vision & defect detection: Classifying images (e.g., handwritten digit recognition) or using automated camera systems to identify defective products in industrial manufacturing.



**9. The logic behind "negative mean squared error"**

For error metrics like Mean Squared Error (MSE), a lower value represents better performance (0 error is optimal).  Because scikit-learn requires that higher scores always represent better performance, it negates the metric by multiplying it by $-1$.  Example: If Model 1 has an MSE of $100$ and Model 2 has an MSE of $150$, Model 1 is better. Multiplying by $-1$ yields $-100$ and $-150$. Because $-100 > -150$, scikit-learn correctly identifies Model 1 as having the higher, better score, maintaining the correct ranking
