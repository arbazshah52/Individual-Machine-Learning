## Faktafrågor (Factual questions)

## 1. 
1. **Definiera problemet och skapa en helhetsbild** (Define the problem and get an overview, p. 56): understand what you want to achieve and build an overall picture. Questions to ask: What is the goal of the project? If the goal is met, will what we build actually be useful? Are there existing solutions or "workarounds" we could use instead? How do we measure the project's progress? Which people are good to involve in the project and discuss it with?

2. **Få tillgång till datan** (Get access to the data, p. 56): data is of central importance for building ML models, so a first step is to get access to it. Questions to ask: What data do we need for the project and how do we get access to it? Do we handle the data lawfully (e.g. protected data that needs anonymizing, handled in a legally correct way)? If the data will be used for ML modeling, a dataset should be set aside here as test data, which must not be used until the very end; training data and validation data are also created here (or, alternatively, only training and test data if k-fold cross-validation will be used later). An info box on this page ("Finns datan för det vi ursprungligen ville göra? En iterativ arbetsprocess.") notes that at this step we might discover that what we originally wanted to do isn't possible because the data doesn't exist — which can send us back to step 1 to reconsider the goal, or to start collecting the eeded data instead.

3. **Utforska datan, gör en EDA** (Explore the data, do an exploratory data analysis, p. 57): explore and analyze the data through an EDA to understand what kind of data you're working with — computing means, medians, modes and similar, and creating visualizations to understand relationships and spot patterns. This should be done on a copy of the data so the original is not accidentally altered, and EDA is performed on the training data. Questions to ask: Are there patterns visible in the data? What do they say, what might they depend on? Are there visualizations that could be especially useful for understanding the data and discovering patterns? Does anything look wrong in the data, and how can we verify whether it actually is wrong? Are there missing values in the data? An info box here ("Behövs ML-modellering eller räcker en EDA?") notes that in reality many business questions can be answered with an EDA alone, without any ML modeling being needed at all — so it's worth asking "is ML modeling actually required for what we want to achieve?"

4. **Bearbeta datan** (Process the data, p. 58): again work on a copy of the data. When transforming the data it's a good idea to write functions for the transformations, since the same transformations can then be reapplied when new data arrives, and the transformations can even be treated as a hyperparameter (e.g. to be optimized with GridSearch). Questions to ask: Are there outliers in the data, and how should they be handled (data cleaning)? Are there missing values, and how should they be handled (e.g. removing rows or filling with the mean/median)? Is there data/variables that are not useful for our purpose and can be removed? Should feature engineering be done (creating new variables or transforming existing ones) — this is usually done here, though in practice there isn't a sharp boundary between this step and ML modeling, and one often works iteratively between the two.

5. **ML-modellering** (ML modeling, p. 59): it's reasonable to start by trying several different models with default hyperparameter values, to get a sense of which type of model seems to perform well; with experience one builds intuition for which models work for which problems, but ultimately the models are evaluated on validation data to see which actually perform well. From there, hyperparameters are tuned further for the promising models, and different variable sets can be tried (variable selection) — this is an iterative process. Eventually you end up with the model that performs best on the validation data; that model can start being used once its performance is good enough and meets requirements, but before that its generalization ability should finally be evaluated on the test data, to get an idea of how well it can be expected to perform once used in production. An info box here ("Överanpassa inte modellen till testdatan") warns explicitly against adjusting the final model based on its test-data performance — more on this under Question 5 below.

6. **Presentera din lösning för intressenter** (Present your solution to stakeholders, p. 60): often the solution needs to be presented to different stakeholders — colleagues on a team, marketers, managers, or project leaders. Think about who the target audience is and what information is useful to them. Points to consider: which level of abstraction to use (a technically oriented audience may be interested in modeling details, while a leadership group is often more interested in concrete results, actions the work could lead to, and economic consequences); before presenting, think through what the message is and how to communicate it effectively; visualizations of different kinds are often helpful.

7. **Produktionssättning av modellen och övervakning av implementeringen** (Putting the model into production and monitoring the implementation, p. 60): if a model is successfully built and actually put into production, it's good to think about: creating various unit tests to ensure the functionality is as intended; monitoring the model's performance, since models can degrade over time as the data starts to look different due to changes in innovation, the economy, and people's behavior; models are therefore often retrained regularly on new data, for example monthly or weekly.

**Is this followed as a straight progression, or worked iteratively?** The book answers this directly at the start of Section 2.1 (p. 55): the checklist has seven steps and it is easy to get the impression that these steps are carried out in a straight progression — step one, then step two, then step three, and so on. The book explicitly says this is *not* how it works: "Ofta kan vi behöva hoppa mellan stegen, exempelvis för att vi fått en insikt som påverkar vår analys från föregående steg eller för att det vi egentligen ville göra inte är möjligt." (Often we may need to jump between the steps, for example because we've gained an insight that affects our analysis from a previous step, or because what we originally wanted to do turns out not to be possible.) So in reality one is prepared to work **iteratively**, going "back and forth" and jumping between the steps. The book also stresses that the checklist is meant as a support for the work, and is nothing we either need to, or should, follow slavishly.

This is explained directly in the "MLOps" info box in Section 2.1.7 (p. 62). The book states plainly: "När vi säger att vi produktionssätter en modell menar vi helt enkelt att den börjar användas." (When we say that we put a model into production, we simply mean that it starts being used.) This can happen in different ways, and the book gives several examples:

- A model can be used on a website — this is the case for chatbots such as ChatGPT, which are accessed via a website.
- A model can be used in applications such as social media, where ML models recommend content in a way that makes users spend more time in the app.
- A model can be used in various devices such as cameras, for example to detect defects in an industrial production process, or in self-driving cars. Generally speaking, ML models are usable within the area known as the **Internet of Things (IoT)** (in Swedish, *sakernas internet*).
- A model can be used to create a new column of predicted values in an SQL database. This column of predicted values can then be used for different purposes — for example, if a model predicts the probability that a customer will churn, a company can put measures in place, such as targeted marketing, for all customers who, say, have over 35% predicted risk of churning.

### What is scikit-learn?

Section 2.4 (p. 90) describes scikit-learn as an **open source machine learning library for Python**. The library includes, among other things, algorithms for regression, classification, clustering, model selection, and data processing. Its first public release was in 2010, and it is today a very popular and widely used library within ML, both in industry and in academic contexts. The book illustrates this with Figure 2.2, a 2020 survey of professionals in the data-science field showing which ML libraries are used the most — scikit-learn was the most popular, used by 82.8% of respondents (with TensorFlow and Keras tied for second/third place at 50.5% each). The name *scikit-learn* comes from the project's original aim to be a "scientific toolkit for machine learning."

### Design principles (Section 2.4.1, pp. 92–94)

The book names four central design principles that scikit-learn follows: **consistency**, **inspection**, **limitation of classes**, and **sensible default hyperparameter values**.

- **Consistency (Konsistens)**: all objects should have a consistent interface and a limited set of methods (methods are simply functions belonging to classes), and the documentation is consistent as well, following the same pattern throughout. The book demonstrates this with code: training a linear regression model and a decision tree — two completely different models — involves exactly the same process: (1) instantiate the model, (2) train it with `.fit()`, and (3) use it to make predictions with `.predict()`.
- **Inspection (Inspektion)**: constructor parameters (hyperparameters — called "parameters" in the scikit-learn documentation) are accessed with the `.get_params()` method. Learned parameters (called "attributes" in the scikit-learn documentation) are stored and accessible as public attributes, accessed by using a trailing underscore as a suffix (e.g. `.intercept_`, `.coef_`). An info box on p. 94 clarifies a potentially confusing point: in scikit-learn's own terminology, hyperparameters are called "parameters" and (learned) parameters are called "attributes" — but generally within ML, hyperparameters are the things that answer "how do we learn", and parameters are the things that answer "what have we learned" (as established already in Chapter 1).
- **Limitation of classes (Begränsning av klasser)**: only model algorithms, such as `LinearRegression`, are represented as purpose-built classes. Data is represented as NumPy arrays or SciPy sparse matrices, and hyperparameters as ordinary Python strings or numbers whenever possible.
- **Sensible default values for the hyperparameters (Rimliga standardvärden för hyperparametrarna)**: whenever values need to be specified for hyperparameters, scikit-learn provides sensible default values, meaning a typical user doesn't need to think about setting them. This makes the library easy to use and implement, and also lets more people actually work with ML without necessarily understanding theoretical details that, in many cases, have little practical significance.

### Estimators, predictors, transformers (Section 2.4.2, pp. 95–96)

- **Estimators**: the central object in scikit-learn is called an *estimator*. Estimators learn ("estimate") from data through the `.fit()` method. Estimators are first (1) instantiated, and only then are they (2) trained using `.fit()`.
- **Predictors**: estimators that can make predictions through a `.predict()` method are called *predictors*. Predictors are therefore a subset of estimators — for example, `LinearRegression()` is both an estimator and a predictor. Predictors must always have a `.score()` method that quantifies how good the predictions being made are; as established already in Chapter 1, a higher score is always better in scikit-learn.
- **Transformers**: estimators that can transform data through a `.transform()` method are called *transformers*. Transformers are therefore also a subset of estimators — for example, `StandardScaler()` is both an estimator and a transformer. The `.transform()` method takes the data to be transformed as an argument and returns the transformed data. All transformers in scikit-learn also have a convenience method, `.fit_transform()`, which performs both fitting and transforming in a single step.

## 6. Many AI/ML projects do not reach their originally set goals, or don't even pass some form of prototype stage. What do you think this is due to, and how should we relate to it?

The book addresses this directly in the info box "De flesta AI/ML-projekten misslyckas" (Section 2.1, p. 62), right after the seven-step checklist and just before the worked example begins. It states plainly that most AI/ML projects do not reach their originally set goals, or don't even get past some form of prototype stage, and that this can be due to reasons such as **lack of the necessary data** ("brist på behövlig data"), **poor models** ("dåliga modeller"), or **the right competence and resources being lacking** ("rätt kompetens och resurser saknas"). It also cites circulating estimates that around **85%** of all ML projects fail: "Det cirkulerar olika uppskattningar om att cirka 85% av alla ML-projekt misslyckas."

Section 2.3, "Utmaningar inom ML" (pp. 90–91), lists several concrete, recurring challenges that plausibly explain a large part of this failure rate, and almost all of them trace back to data:

- **Too little data**: many application areas require thousands of observations for a model to be trained well enough to perform adequately; in some cases, such as image analysis, millions of images may be required. No matter how skilled the modeler, good results are hard to achieve without sufficient data.
- **Non-representative data**: the data used to train models generally needs to be representative of the data the model will later be used to predict on, or the model won't perform well in practice.
- **Poor data quality**: if the data has errors — for example because it was carelessly collected or incorrectly registered — ML models will struggle to learn the underlying patterns. Missing values are also common and need to be handled (e.g. removing rows, or filling in with the mean/median), and this handling itself is a choice that can affect the outcome.
- **Irrelevant features**: as the book puts it, "shit in – shit out" — without good features, it generally isn't possible to build good ML models, regardless of how sophisticated the models themselves are.
- **Overfitting and underfitting**: there is always a risk of over- or under-fitting a model when building it, which can make it perform worse than expected once used on new data.

**How we should relate to it**: the book's own answer, in that same info box, is to avoid treating this in purely binary terms of a project "succeeding" or "failing": "Undvik därför att tänka för binärt i termer av att projekt 'lyckas' eller 'misslyckas' och försök istället se hur det som gjorts kan vara användbart eller vilka lärdomar som tagits och hur det kan komma till nytta för framtiden." (So avoid thinking too binarily in terms of projects "succeeding" or "failing", and instead try to see how what was done can be useful, or what lessons were learned and how they can be put to use for the future.) The book gives its own example of this reframing: even if the original goal was to build an ML model, and that specific model never gets used, the EDA that was carried out along the way might still be useful on its own.

Building on the book's own framing: given that it is realistic to expect that not every project we work on will reach production, and that most of the concrete challenges in Section 2.3 are fundamentally about data (availability, representativeness, and quality), a sensible way to relate to this is to invest properly in the "Get access to the data" and "Explore/process the data" steps of the checklist early on, set realistic expectations about the odds of full success from the outset, and judge a project's value not only by whether the original model reaches production, but also by the insights, reusable pipeline, or lessons it produces along the way — exactly as the book's own info box recommends.

*Source: Section 2.1, p. 62, info box "De flesta AI/ML-projekten misslyckas"; Section 2.3, pp. 90–91, "Utmaningar inom ML".*
