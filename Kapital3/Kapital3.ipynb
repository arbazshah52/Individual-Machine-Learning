### 1. What characterizes regression problems? Give some examples of application areas.

> "Regression problems are about predicting a continuous dependent variable (y) using
> independent variables (x)." *(translated; original: "Regressionsproblem handlar om att
> predicera en kontinuerlig beroende variabel (y) med hjälp av oberoende variabler (x).")*

So what characterizes a regression problem is that **the target variable (y) is
continuous** (it can take any numeric value within a range), as opposed to classification
problems (Chapter 4) where the target variable is categorical. The book gives as an
example that "it could, for instance, be about predicting a person's salary (y) based on
the person's age (x)" (p. 105-106).

**Examples of application areas (p. 106, the book's own examples)**

- **Car valuation** - Predicting the sale price of a car based on variables such as
  mileage, age, fuel type and brand.
- **Real estate valuation** - Predicting the sale price of a property based on variables
  such as size, year built, geographic location and number of rooms.
- **Agricultural production** - Predicting how large a harvest (e.g. potatoes) will be in
  kilograms, based on variables such as temperature, amount of rainfall and fertilizer use.
- **Insurance** - Predicting the expected claim cost (in currency) for a customer based on
  variables such as age, education, geographic location and prior claims history.
- **Product demand** - Predicting how many units of a product will be sold (volume) based
  on price, season, campaigns and past sales data.
- **Salary level** - Predicting a person's salary based on variables such as education
  level, age, experience, industry and geographic location.

### 2. Explain the evaluation metrics RMSE, MSE and MAE.

The book covers these three metrics in Section 3.2 (p. 106-108), noting that they are
"common" metrics for evaluating regression models.

**Root Mean Squared Error (RMSE)** (Section 3.2.1, p. 106)

> "RMSE is the metric used most extensively when evaluating regression models."

Formula (Equation 3.1):

$$\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$

**Mean Squared Error (MSE)** (Section 3.2.2, p. 107)

> "Generally speaking, we use RMSE when we want to be able to interpret how large an error
> a model makes. We can also use it to rank different models [...] However, if we are only
> interested in ranking models, it is generally more efficient to use MSE, since the
> ranking of the models will be the same, but MSE is computationally cheaper because no
> square-root calculation needs to be performed."

Equation 3.2:

$$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

In other words, MSE is the same as RMSE but without the square root. The book demonstrates
this with `sklearn.metrics.mean_squared_error`: with `y_true = [3, -0.5, 2, 7]` and
`y_pred = [2.5, 0.0, 2, 8]`, the result is `0.375`.

**Mean Absolute Error (MAE)** (Section 3.2.3, p. 107-108)

> "Just as for RMSE and MSE, a lower value for MAE also indicates a better model."

Equation 3.3:

$$\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i|$$

where $|x|$ is the absolute value (Equation 3.4). The same example data gives `0.5` with
`mean_absolute_error`.

**The difference between the metrics - the book's own proof (p. 108)**

> "It can be mathematically proven that RMSE ≥ MAE, since RMSE places a larger penalty on
> large deviations than MAE does. This means that in cases where we want to avoid large
> errors, RMSE is a suitable metric to use. In cases where all errors are valued linearly
> with respect to the size of the error, MAE is a suitable metric to use."

The book illustrates this concretely: true values `y_true = [100, 80]`, and two competing
models predicting `[90, 75]` and `[90, 15]` respectively. The result:

| | Model 1 | Model 2 | Factor |
|---|---|---|---|
| MAE | 7.5 | 37.5 | ×5 |
| RMSE | 7.906 | 46.503 | ×~5.89 |

The book's conclusion: "MAE thus increased linearly while RMSE increased non-linearly,
meaning the latter metric places a larger penalty on large deviations." Finally: "Generally
speaking, RMSE is the metric used the most, but depending on what one wishes to achieve,
MAE can also be a suitable metric to use" (p. 108).


### 3. If we want to rank different models, does it matter whether RMSE or MSE is used? Why?

**No, it does not matter for the ranking** - the book answers precisely this question
explicitly in Section 3.2.2 (p. 107):

> "However, if we are only interested in ranking models, it is generally more efficient to
> use MSE, since the ranking of the models will be the same, but MSE is computationally
> cheaper because no square-root calculation needs to be performed, which can save time."

The reason (beyond the book's direct statement) is mathematical: since
$\text{RMSE} = \sqrt{\text{MSE}}$ and the square-root function is strictly increasing
(monotonic) for all non-negative numbers, if Model A has a lower MSE than Model B, Model A
is guaranteed to also have a lower RMSE than Model B - the ranking between models can never
be reversed by taking the square root. So:

- **Ranking (which model is best)**: identical regardless of whether RMSE or MSE is used.
- **Interpreting the size of the error**: RMSE is nevertheless preferable when you want to
  understand/communicate how large the actual error is, since RMSE has the same unit as
  $y$ (the target variable), while MSE has the squared unit (p. 107: "Generally speaking,
  we use RMSE when we want to be able to interpret how large an error a model makes").
- **Computational cost**: MSE is cheaper to compute since the square root doesn't need to
  be calculated, which the book highlights as the practical advantage of MSE when you are
  only ranking models (e.g. inside a `GridSearchCV` search over many hyperparameter
  combinations).

### 4. Explain, very briefly, what gradient descent is.

The book introduces gradient descent in Section 3.3.2 (p. 122-127), in connection with how
the parameters $\hat{\theta}_0, \hat{\theta}_1$ of the linear regression model are
estimated.

> "Gradient descent is an optimization algorithm used to find the minimum or maximum of a
> function. [...] The intuition behind the algorithm is that when we want to find the
> minimum of a function, we move in the direction with a negative slope and continue until
> the slope is 0." (p. 123)

Concretely, training the model means finding the $\hat{\theta}_0, \hat{\theta}_1$ that
minimize the cost function (loss function/cost function), which for linear regression is
MSE (Equation 3.12, p. 122):

$$\text{MSE} = \frac{1}{n}\sum_{i=1}^n(y_i-\hat{y}_i)^2$$

According to Figure 3.3 (p. 124), the algorithm works in four steps:
1. **Random initialization** - you start somewhere (a random point on the curve).
2. You move in the direction where the slope (derivative) is negative, with a step size
   called the **learning step/learning rate**.
3. This is repeated a number of times, called the **number of iterations**.
4. You stop when the slope is 0, which is the minimum point (the optimal parameter
   $\hat{\theta}$).

The book points out (p. 123) that for linear regression, MSE is a **convex function**,
which guarantees that the local minimum found is also the global minimum. In general
(for other, more complex models), however, there are risks: getting stuck in a **local
minimum** (Figure 3.4), getting stuck on a **plateau**, or taking too-large steps (too
high a learning rate) so that the algorithm "bounces" past the optimum and never converges
(Figures 3.5-3.6, p. 125-126). The book also mentions three variants: *batch*, *stochastic*
and *mini-batch gradient descent* (p. 127, Figure 3.7), and notes that `StandardScaler`
(scaling the variables) makes it easier for the algorithm to reach the optimum faster
(Figure 3.8, p. 128).

The name comes from the fact that the algorithm tries to find the point where the
**gradient** (a generalization of the derivative for functions of several variables) is
zero (p. 128).


### 5. What is the bias-variance trade-off? Why aren't more complex models always better?

The book covers this in Section 3.3.3 (p. 128-129):

> "The bias-variance trade-off states that the error a model makes when predicting new,
> unseen data can be divided into three parts."

The three parts (p. 128):

1. **Bias** - "This error is due to incorrect model assumptions. For example, we might
   assume the data is linear when in reality it is quadratic. If we then use an ordinary
   simple linear regression model [...] instead of a polynomial regression with quadratic
   terms [...] we get bias. A model with high bias can lead to underfitting."
2. **Variance** - "This error is due to the model being sensitive to fluctuations in the
   training data when it is trained. For example, very flexible models, such as a
   high-degree polynomial regression, risk having small changes in the data lead to the
   model looking very different, which gives it high variance. A model with high variance
   can lead to overfitting."
3. **Irreducible error** - "This is error that is simply due to randomness in the data;
   such errors are irreducible and nothing we can directly do anything about."

**Why aren't more complex models always better?**

The book explains (p. 129):

> "When we increase a model's complexity, for example going from ordinary simple linear
> regression to polynomial regression, we generally increase its flexibility but decrease
> its bias, and vice versa: when we decrease a model's complexity, we generally decrease
> its variance but increase its bias. That is why a more complex model isn't always
> better - the increase in variance can cost more than what the decrease in bias is worth.
> That is the reason it is called a trade-off."

In other words: a more complex model (e.g. a high-degree polynomial regression) can reduce
bias (it can capture more complicated patterns), but this often comes at the cost of a
sharply increased variance (the model becomes sensitive to the specific training data and
generalizes worse to new data). If the increase in variance is larger than the decrease in
bias, the total error on **new, unseen data** actually gets worse, even though the model is
more complex and fits the training data better. This is illustrated in Figure 3.9
(p. 130), where a flexible polynomial model (blue) is described as having "higher variance
and lower bias compared to the red model [simple linear regression]", and the book notes
that "to know which model is actually better, we would have needed to evaluate them on
validation data" - which is exactly why train/val/test splits are used rather than just
looking at the training error. This result is also the foundation for why regularization
(ridge, lasso, elastic net) is introduced - to deliberately increase bias a little in order
to decrease variance more, as the book states (p. 129): "The price we pay for the lower
variance is generally higher bias, but we hope the decrease in variance ends up being
greater than the increase in bias."


### 6. Explain, in broad terms, how each regression model works

#### a) Linear regression (Section 3.3.1, p. 111-121)

Linear regression is one of the most central models. In its simplest form (**simple linear
regression**, one independent variable), predictions are made according to Equation 3.5:

$$\hat{y} = \hat{\theta}_0 + \hat{\theta}_1 x$$

where $\hat{\theta}_0$ (intercept) and $\hat{\theta}_1$ (slope) are estimated parameters.
The difference between the true value and the predicted value is called the **residual**,
$e = y - \hat{y}$. This is illustrated in Figure 3.1 (p. 113, see also the answer to
Reasoning Question 9 below).

The model generalizes to **multiple linear regression** with several independent
variables (Equation 3.7, p. 117): $\hat{y} = \hat{\theta}_0 + \hat{\theta}_1 x_1 +
\hat{\theta}_2 x_2 + \dots + \hat{\theta}_p x_p$, which in matrix form is written
$\hat{\mathbf{y}} = \mathbf{X}\hat{\theta}$ (Equation 3.10, p. 119).

The model can also be extended with **polynomial terms** (e.g. $x^2$) via
`PolynomialFeatures`, giving **polynomial regression** (Equation 3.6). The book stresses
in an info box (p. 118) that polynomial regression is still considered a *linear* model,
since it is linear with respect to the parameters $\hat{\theta}_i$, even though it is not
linear with respect to $x$.

The parameters are estimated by minimizing the cost function MSE, which is done using
**gradient descent** (see Question 4). The book also shows (info box, p. 121) that the
same model is used both in classical statistics (for inference/understanding
relationships) and in ML (for making predictions on new data) - "the model itself is the
same" but the purpose differs.

#### b) Ridge regression - L2 regularization (Section 3.3.4, p. 129-132)

> "Ridge regression is a regularized variant of the linear regression model, in which
> overfitting is counteracted by shrinking the value of the weights $\hat{\theta}_i$. By
> shrinking the weights and thereby constraining the model, it becomes less flexible,
> which increases bias but decreases variance."

The cost function (Equation 3.13) adds a penalty term based on the square of the weights:

$$J(\theta) = \text{MSE}(\theta) + \alpha\frac{1}{2}\sum_{i=1}^{p}\theta_i^2$$

where $\alpha$ (the regularization strength) controls how hard the weights are shrunk
toward 0 - the higher $\alpha$, the less flexible/more regularized the model becomes
(Figure 3.10, p. 132). A suitable value for $\alpha$ is typically chosen using `grid
search` (demonstrated with `sklearn.linear_model.Ridge` + `GridSearchCV`).

#### c) Lasso regression - L1 regularization (Section 3.3.5, p. 133-135)

> "Lasso (*Least Absolute Shrinkage and Selection Operator*) is a regularized variant of
> linear regression."

The cost function (Equation 3.14) instead penalizes the absolute value of the weights:

$$J(\theta) = \text{MSE}(\theta) + \alpha\sum_{i=1}^{p}|\theta_i|$$

The important difference versus ridge: "it can be mathematically proven that introducing
[the penalty term] tends to set the weights of less important variables to **0**. [...]
This can be compared to ridge regression, where the weights are generally shrunk so that
they approach 0." This means lasso can automatically perform **feature selection** -
useful for high-dimensional datasets where some variables are suspected to be unimportant.

#### d) Elastic net (Section 3.3.6, p. 135-136)

> "Elastic Net is a combination of lasso (L1) and ridge (L2) regularization. It balances
> both techniques using a *mix ratio*, r."

Cost function (Equation 3.15):

$$J(\theta) = \text{MSE}(\theta) + r\alpha\sum_{i=1}^{p}|\theta_i| + \frac{1-r}{2}\alpha\sum_{i=1}^{p}\theta_i^2$$

When $r=0$, the cost function becomes identical to ridge regression, and when $r=1$ it
becomes identical to lasso regression. "This combination allows Elastic Net to select
important features (like lasso) while still retaining some small, non-zero weights (like
ridge)." The hyperparameters `alpha` and `l1_ratio` (corresponding to $r$) are typically
tuned with grid search.

#### e) Support vector machines (SVM) (Section 3.3.7, p. 136-139)

> "For regression problems, the SVM model tries to create a road, of a given width, that
> contains as many observations as possible. The center line of the road becomes the
> model's predicted values, $\hat{y}$."

The width of the road is controlled by the hyperparameter **epsilon ($\epsilon$)**:
observations inside the road (prediction error smaller than $\epsilon$) are ignored by the
model during training - the model is said to be "$\epsilon$-insensitive" (Figure 3.12,
p. 137). A lower $\epsilon$ gives a more flexible model; a higher $\epsilon$ gives a less
flexible model. SVM models should generally be standardized (e.g. via `StandardScaler`)
and are best suited to small/medium-sized datasets. The book also mentions the **kernel
trick** (covered in more depth in the classification chapter, Section 4.3.2) - it produces
the same effect as adding transformed variables (e.g. polynomial terms) without actually
having to create them, saving computation time. scikit-learn provides `LinearSVR` (no
kernel) and `SVR` (with a kernel).

#### f) Decision trees (Section 3.3.8, p. 139-143)

A decision tree makes predictions by traversing the tree from the root node, through
internal nodes, to a leaf node:

> "The model will traverse the tree from the topmost node, called the *root node*, and
> continue through *internal nodes* to finally reach the outermost nodes, called *leaf
> nodes*."

At each internal node a question is asked about an independent variable and a threshold
value (e.g. "is $x_2 \le 0{.}438$?"), and depending on the answer (true/false) you move to
the left or right child node (see Figure 3.13, p. 140, and the answer to Reasoning
Question 10 below). "How does the model choose which independent variable and threshold
value to use at each question? The answer is that it chooses the variable and threshold
value that make MSE (*squared error*) as low as possible at each step" - which
`scikit-learn` handles automatically.

Decision trees are described as **white box models** (see Question 7) because it is easy
to follow and understand how an individual prediction is made. They should generally be
regularized (e.g. with `max_depth`, `min_samples_split`, `min_samples_leaf`,
`max_leaf_nodes`) to avoid overfitting (Figure 3.15, p. 143).

#### g) Ensemble learning (Section 3.3.9, p. 144-149)

> "Ensemble learning is about combining several models in the hope that, together, they
> can perform better than each individual model could on its own."

The book covers two variants:

- **Voting regression**: several different models are trained (e.g. linear regression,
  decision tree, SVM) on the same dataset, and the final prediction is the **average** of
  the individual models' predictions (Figure 3.16, p. 145). Can be done manually or with
  `sklearn.ensemble.VotingRegressor`.
- **Bagging and pasting**: several *new* datasets are created by taking random samples from
  the original dataset - with replacement (**bagging**, the same observation can appear
  multiple times) or without replacement (**pasting**, the same observation cannot appear
  multiple times). Then *one and the same type of model* is trained on each of these
  datasets, and the final prediction is again the average of the individual models'
  predictions (Figure 3.17, p. 146). Done with `sklearn.ensemble.BaggingRegressor`
  (controlled with `bootstrap=True/False`, `n_estimators`, `max_samples`, `max_features`;
  see Question 8 for more detail).

#### h) Random forest (Section 3.3.10, p. 149-150)

> "Random forest is an *ensemble* of decision trees that is usually created through
> *bagging*, although *pasting* also occurs."

In other words, random forest is a special case of bagging where the base components are
always decision trees. It could in principle be implemented manually with
`BaggingRegressor` and `DecisionTreeRegressor`, but scikit-learn has its own dedicated
class: `sklearn.ensemble.RandomForestRegressor`. Just as with individual decision trees,
it is "often good to regularize random forest" (e.g. via `max_depth`, `n_estimators`,
`min_samples_split`, tuned with `GridSearchCV`). Random forest is described in the book as
a **black box model** (unlike the individual decision tree), see Question 7.


### 7. What is meant by white box models and black box models?
- **White box models** - models where you can easily, transparently, and step-by-step
  understand/follow *why* the model made a certain prediction. The book gives decision
  trees as an example: "For example, with a decision tree we can manually step through it
  and see which prediction is made."
- **Black box models** - models whose internal computations can certainly be followed
  mathematically, but where it is difficult to explain in an intuitive/understandable way,
  in words, why the model arrived at a certain prediction. The book mentions random forest
  and neural networks as examples: "We generally cannot do that with random forest and
  neural networks."


## 8.
- **Bagging** (*Bootstrap Aggregating*): "With bagging, we perform a random sample **with
  replacement** (meaning the same observation can appear multiple times in the new dataset
  that is created)."
- **Pasting**: "With pasting, we perform a random sample **without replacement** (meaning
  the same observation *cannot* appear multiple times in the new dataset that is
  created)."
  
  In scikit-learn's `BaggingRegressor`, this is controlled with the `bootstrap` parameter:
`bootstrap=True` gives bagging (the default), while `bootstrap=False` gives pasting

### 9. Explain how Figure 3.1 on page 113 can be interpreted.

**Figure 3.1** (p. 113) shows "a schematic picture of the simple linear regression model",
with $y=$ Income on the y-axis and $x=$ Age on the x-axis. The caption reads:

> "The black dots represent the data ($y_i$), and the blue line is the predicted values,
> $\hat{y}=\hat{\theta}_0+\hat{\theta}_1x$, according to our regression model. The red
> vertical lines are the residuals, $y_i-\hat{y}_i=e_i$."

The interpretation is:

- **The black dots** are the actual (observed) data points - in the example, people's
  actual income at a given age.
- **The blue line** is the model's predictions, $\hat{y}=\hat{\theta}_0+\hat{\theta}_1x$ -
  a straight line (compare $y=kx+m$, p. 111) that the model has fit to the data.
  - $\hat{\theta}_0$ (the intercept) is shown in the figure as where the line crosses the
    y-axis, i.e. the predicted value when $x=0$.
  - $\hat{\theta}_1$ (the slope) is shown in the figure as the line's angle of slope and
    "answers the question 'how much does income increase when we increase x (age) by one
    unit'" (p. 112).
- **The red vertical lines** are the residuals $e_i=y_i-\hat{y}_i$, i.e. the vertical
  distance between each observed data point and the line's predicted value for the same
  $x$-value - in other words, the model's error/deviation for that particular observation.

Overall, the figure shows how the simple linear regression model tries to draw a straight
line through the cloud of data points such that the distance (the residuals) between the
line and the points is as small as possible (in terms of MSE, see Questions 2 and 4) -
the shorter the red lines are in total, the better the model.

### 11. The coefficient of determination (R²)

The book itself, in Chapter 3, only covers RMSE, MSE and MAE (Section 3.2) as evaluation
metrics for regression. As the question itself instructs, R² (the coefficient of
determination) is instead covered here using scikit-learn's own documentation
(`sklearn.metrics.r2_score`, and the regression-metrics section of the "Model evaluation"
user guide).

**What the metric is**

R² (the *"coefficient of determination"*) measures how much of the variance in the target
variable ($y$) the model manages to explain, compared to a trivial baseline model that
always predicts the mean of $y$. According to scikit-learn's documentation for
`r2_score`, it is calculated as:

$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}} = 1 - \frac{\sum_i(y_i-\hat{y}_i)^2}{\sum_i(y_i-\bar{y})^2}$$

where $SS_{res}=\sum_i(y_i-\hat{y}_i)^2$ is the residual sum of squares (compare
$\text{MSE}\times n$, Question 2) and $SS_{tot}=\sum_i(y_i-\bar{y})^2$ is the total sum of
squares, i.e. the variance of $y$ around its mean $\bar{y}$.

**Interpretation (per scikit-learn's documentation)**

- $R^2 = 1{.}0$: best possible value - the model predicts perfectly, all variance in $y$
  is explained by the model.
- $R^2 = 0{.}0$: the model performs exactly as well as a constant model that always
  predicts the mean $\bar{y}$ - i.e. the model has no predictive value beyond just
  guessing the mean.
- $R^2 < 0$: the model performs **worse** than just guessing the mean (this can happen
  since a model can be "arbitrarily worse" than the baseline, unlike e.g. RMSE which is
  always $\ge 0$).

The documentation also notes that `r2_score` by default
(`multioutput='uniform_average'`) can handle several target variables at once, that R² is
scikit-learn's **default metric** (the `.score()` method) for regression models (e.g.
`LinearRegression`, `Ridge`, etc.), and that it is available as `scoring='r2'` in
functions such as `cross_val_score` and `GridSearchCV`.

**Relation to RMSE/MSE (Questions 2-3)**

Unlike RMSE/MSE/MAE, which are error metrics where **lower is better** and which are
scale-dependent (measured in the same unit as $y$, or its square), R² is a **normalized,
scale-free** metric between (in principle) $-\infty$ and $1{.}0$ where **higher is
better**. Since R² is an (affine, monotonically decreasing) transformation of MSE - for a
given dataset, $R^2 = 1 - \text{MSE}/\text{Var}(y)$ - R² gives the same ranking of models
as MSE/RMSE (compare the reasoning in Question 3: if MSE decreases, R² increases, so a
model with lower MSE always has a higher R² on the same data). The advantage of R² is that
it is easier to interpret "how good" a model is in relative terms (e.g. "the model
explains 92% of the variation in price") without needing to know the scale of $y$, while
RMSE/MAE give a more direct, concrete measure of the expected error size in $y$'s own
unit.
