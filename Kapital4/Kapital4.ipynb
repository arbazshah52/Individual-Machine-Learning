### 1. What characterizes classification problems? Give some examples of application areas.

The book defines classification problems in Section 4.1 (p. 151):

> "Classification problems are about predicting categorical data, where the data can take
> two or more classes."

So what characterizes a classification problem is that **the target variable is
categorical** (it belongs to one of a fixed set of classes), as opposed to regression
problems (Chapter 3) where the target variable is continuous.

**Examples of application areas (p. 151-152, the book's own examples)**

- **Customer analysis** - Predicting whether a customer will *churn* (i.e. leave the
  company) or not, based on variables such as number of purchases, age, gender and number
  of support tickets.
- **Healthcare** - Predicting whether a patient has a certain disease or not (sick or
  healthy) based on medical data such as blood pressure, pulse, etc.
- **Image recognition** - Identifying objects in an image based on variables such as
  color, shape and texture - for example identifying pest damage in agricultural fields or
  on golf courses.
- **Anomaly detection** - Used, among other things, in manufacturing processes to
  automatically identify defective products using images from cameras.
- **Finance** - Predicting whether a stock will rise in value or not, based on e.g. company
  information such as profitability and revenue, and macroeconomic data such as inflation
  and interest rates.
- **Credit risk** - Assessing whether a bank customer should be granted a mortgage or not,
  based on data such as income, age, marital status and any payment remarks.

**Types of classification (p. 152-153)**

The book distinguishes three types:
- **Binary classification** - exactly two classes, e.g. {churn, stay} or {1, 0}.
- **Multiclass classification** - more than two classes, e.g. predicting which animal is
  shown in an image out of {dog, cat, horse, bird}.
- **Multioutput (multiclass) classification** - predicting two or more output variables at
  the same time, e.g. simultaneously predicting whether a person is over 18 {yes, no} and
  which gender they are {man, woman} (multioutput), or, if at least one output can take
  more than two classes (e.g. also predicting which of {southern Sweden, central Sweden,
  northern Sweden} the person lives in), *multioutput multiclass classification*. The book
  notes it does not go deeper into these last two types and focuses on binary and
  multiclass classification.


### 2. Explain how the OvR and OvO algorithms work.

The book explains this in Section 4.1.1, "OvR- and OvO-algoritmerna" (p. 154):

> "Binary classification models can, with the help of two algorithms called One-vs-Rest
> (OvR) and One-vs-One (OvO), be extended to also be able to handle multiclass
> classification."

**One-vs-Rest (OvR)**

> "OvR means that for each class in a multiclass problem, a model is trained to be able to
> distinguish whether a data point belongs to that class or not. When a new observation is
> to be classified, all the models are run, and the final prediction is the one from the
> model with the highest *score* (such as probability)."

**One-vs-One (OvO)**

> "With OvO we instead train one model per pair of classes. This means that if there are N
> classes, (N × (N-1))/2 models need to be trained. Each model compares a pair of classes
> and predicts one of the two classes. The class predicted the most times overall is
> finally chosen."



### 3. Explain the following evaluation metrics.


#### a) Confusion matrix (Section 4.2.1, p. 155)

> "A confusion matrix is a matrix that visualizes how well a model performs by comparing
> true and predicted values. Each row represents the true class, while each column
> represents the predicted class (it can also be the other way around). The number of
> times each class has been predicted as each respective class by the model is calculated
> and written into the matrix. The diagonal, from the top-left corner to the bottom-right
> corner, shows all the data points that were correctly classified."

For a **binary** classification problem, the matrix has four cells (Figure 4.2, p. 155):
**True Positive (TP)** - the number of times the model correctly predicted a positive
class; **False Negative (FN)** - the number of times the model incorrectly predicted a
negative class (when it was actually positive); **False Positive (FP)** - the number of
times the model incorrectly predicted a positive class (when it was actually negative);
**True Negative (TN)** - the number of times the model correctly predicted a negative
class. For **multiclass** problems (Figure 4.3, p. 156), the matrix extends to one
row/column per class, and the book demonstrates it in scikit-learn with
`sklearn.metrics.confusion_matrix` and visualizes it with `ConfusionMatrixDisplay` (p. 157).

#### b) Accuracy (Section 4.2.2, p. 158)

> "Accuracy is a measure of how large a share of all observations have been classified
> correctly."

Formula (Equation 4.1):

$$\text{Accuracy} = \frac{TP+TN}{TP+TN+FP+FN}$$

The book warns, however (p. 158-159): "A high value for accuracy thus means that a large
share of the observations have been classified correctly. Does this mean our model
performs well? Not necessarily. If our dataset is imbalanced, i.e. one class is
considerably more common than the others, the metric risks being misleading. If a dataset
has data points where 95% belong to class 1 and our model always predicts class 1, the
accuracy score will be 95%, despite the model not being able to predict class 0 at all."
Computed in scikit-learn with `sklearn.metrics.accuracy_score`.

#### c) Precision (Section 4.2.3, p. 159)

> "Precision measures the share of the positive predictions that are actually correct."

Formula (Equation 4.2):

$$\text{Precision} = \frac{TP}{TP+FP}$$

> "Theoretically speaking, if a high value for precision is desired, it is possible to
> build a model that only classifies an observation as positive if it is very certain.
> This will, however, generally often lead to many values that are actually positive being
> incorrectly predicted as negative. This then leads to recall [...] becoming low."

#### d) Recall (Section 4.2.3, p. 159-160)

> "Recall measures the share of the positive class that is predicted correctly."

Formula (Equation 4.3):

$$\text{Recall} = \frac{TP}{TP+FN}$$

> "Theoretically speaking, if we build a model that always classifies an observation as
> positive, we will get 100% recall (assuming there is at least one positive observation
> in the dataset). This does, however, generally lead to precision decreasing, since the
> share of our positive predictions that are actually correct decreases (we always predict
> positive regardless of whether the observation is actually positive or negative)."

Computed in scikit-learn with `sklearn.metrics.precision_score` and
`sklearn.metrics.recall_score` (with `average="macro"` for multiclass problems, i.e.
computing the metric per class and then averaging).

#### e) F1-score (Section 4.2.4, p. 162)

> "F1-score is what is called the harmonic mean of precision and recall."

Formula (Equation 4.4):

$$F_1 = \frac{2}{\text{recall}^{-1}+\text{precision}^{-1}} = 2\cdot\frac{\text{precision}\cdot\text{recall}}{\text{precision}+\text{recall}} = \frac{2TP}{2TP+FP+FN}$$

> "To get a high value for F1-score, both precision and recall need to be high."

The book demonstrates this with a worked table of precision/recall/F1 combinations showing
that F1 is low whenever either precision or recall is low (e.g. precision=1.00,
recall=0.01 gives F1=0.02), confirming that F1 rewards a *balance* between the two rather
than either alone. Computed in scikit-learn with `sklearn.metrics.f1_score`, or all three
metrics at once with `sklearn.metrics.classification_report` (see Reasoning Question 11
below for a full worked example straight from the book).

#### f) The ROC curve (Section 4.2.5, p. 164-166)

> "The ROC curve (from the English *receiver operating characteristic curve*) is similar to
> the precision-recall curve. But instead of visualizing the relationship between precision
> and recall, it visualizes the relationship between TPR, from the English *True Positive
> Rate*, and FPR, from the English *False Positive Rate*."

$$\text{TPR} = \frac{TP}{TP+FN} = \text{recall} \qquad (4.5) \qquad\qquad \text{FPR} = \frac{FP}{FP+TN} \qquad (4.6)$$

Note that TPR is simply another name for recall. The book also introduces *sensitivity*
(= recall/TPR, how well the model finds positive cases) and *specificity* (how well the
model finds negative cases), noting FPR = 1 - specificity, so "the ROC curve visualizes the
relationship between sensitivity and 1-specificity." Just as with precision/recall, the
*threshold* can be adjusted to trade off TPR against FPR (Figure 4.5, p. 165). A "perfect
model" would have a curve going straight up to the point (0,1) - TPR=1, FPR=0. A model that
guesses randomly produces a diagonal line. The area under the ROC curve, **ROC-AUC**
(p. 166), is a single-number summary: 1.0 is a perfect classifier, 0.5 is equivalent to
random guessing, and below 0.5 is worse than random. Computed in scikit-learn with
`sklearn.metrics.roc_auc_score`. The book cautions that "there are situations where the ROC
curve and the associated ROC-AUC metric can be misleading and other evaluation metrics
should be used instead" - for example with a heavily imbalanced dataset where the negative
class dominates, FPR will always stay low even if the model is bad at finding the positive
class.


### 4. What is the precision-recall tradeoff?


> "This phenomenon, that a higher recall generally leads to lower precision and vice
> versa, is called the precision-recall tradeoff."

**How the tradeoff is controlled in practice (p. 161-162)**

Classification models generally work by estimating the probability that a data point
belongs to a class; by default, the model predicts the positive class if that probability
is ≥ 50%. This cutoff is called the **threshold**, and it can be adjusted:

> "If we increase the threshold, only the cases where the model is more certain get
> classified as positive, meaning we get fewer false positives, which in turn leads to
> precision increasing. At the same time, this leads to us missing positive cases, meaning
> there will be more false negatives, which leads to recall decreasing."
>
> "If we decrease the threshold, we allow more data points to be classified as positive,
> which leads to fewer false negatives and thus increases recall. This also leads to
> negative cases incorrectly being classified as positive, meaning we get more false
> positives, which leads to precision decreasing."
