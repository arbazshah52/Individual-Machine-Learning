# Chapter 4 - Classification: Answers to the Review Questions

**Datasets used in the coding exercises:**
- `hr_employee_data.xlsx` (Exercise 13) - the same HR dataset used for the EDA in Chapter 3.
- MNIST handwritten digits (Exercise 15) - since scikit-learn's `fetch_openml("mnist_784")`
  (used in the book's own Code Example 1) could not reach openml.org's API from this
  environment, the identical MNIST data (same 70,000 images, same 28x28 grayscale format,
  same labels) was instead obtained as CSV from the public GitHub mirror
  `phoebetronic/mnist` - the same widely-used raw MNIST split, just distributed as a
  ready-made CSV instead of through the OpenML API.
