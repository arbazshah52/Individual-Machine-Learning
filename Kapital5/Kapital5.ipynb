# Chapter 5 - Dimensionality Reduction: Answers to the Review Questions

## Factual Question 1

**What is meant by the curse of dimensionality?**

**The curse of dimensionality** itself is the name for the fact that, in higher
dimensions, data can behave in ways we neither expect nor are used to from lower
dimensions (p. 218). The book demonstrates this with two concrete examples:

1. **Points become extreme.** If a point is chosen at random inside a 2-dimensional
   unit square, it can be shown mathematically that there is less than a 0.4% chance
   the point lands closer than 0.001 length units to an edge - i.e. it is very
   unlikely to be "extreme" in either dimension. Repeating the same reasoning for a
   100-dimensional unit hypercube gives a probability of just over 18%; for 1,000
   dimensions, over 86%; and for 10,000 dimensions, *"the probability is greater than
   99.999999%"* (p. 218). In other words, in high dimensions almost every point ends
   up being extreme in at least one of its many dimensions.
2. **Points become far apart.** Choosing two random points in a 2D unit square gives
   an average distance between them of about 0.52 length units; in 3D, about 0.66; and
   in a *1,000,000*-dimensional unit hypercube, the average distance becomes about 408
   length units (p. 219). More dimensions simply means more "room", so data points end
   up much farther apart on average. The practical implication: *"if we have a new
   data point that we want to predict, it will on average be further away from the
   data points a model has been trained on, the higher the dimension of the dataset...
   It therefore becomes harder to predict, and the risk of overfitting increases -
   i.e. that the model performs well on the data it was trained on but generalizes
   poorly to new, unseen data"* (p. 219).

## Factual Question 2

**What is dimensionality reduction, and why is it done?**

**Definition.** *"Dimensionality reduction means that we transform a dataset with a
given dimension into a lower number of dimensions. The result is that the dataset ends
up with a smaller number of variables. From a given set of variables, we thus get new
variables that are fewer in number"* (p. 220). 
**Why it is done - the three effects (Section 5.2.1, p. 220):**

1. *"Reduce the time it takes to train a model."* This is described as the main
   motivation - e.g. a model that is retrained every night simply needs to finish
   training within the time available.
2. *"A model trained on dimensionality-reduced data can sometimes achieve better
   predictive ability, even though it is often worse."* The book explains that
   predictive ability commonly gets *worse* after dimensionality reduction (because
   information is lost), which forces a tradeoff between training time and predictive
   ability - its own example: a classification model that takes 15 hours to train and
   reaches 70% accuracy might, after dimensionality reduction, take 8 hours to train
   and reach 65% accuracy; whether that tradeoff is worth it "depends on the case, and
   a tradeoff needs to be made" (p. 220). Occasionally, though, predictive ability
   *improves* after dimensionality reduction - e.g. because unnecessary noise is
   removed, preventing the model from overfitting to random noise instead of genuine
   patterns.
3. *"Being able to visualize data, even though it is not the original data."* If a
   high-dimensional dataset is reduced to two or three dimensions, it becomes possible
   to plot it (as in Figures 5.2 and 5.3, p. 221-222) - keeping in mind that what is
   plotted is a *transformed* version of the data, so a direct interpretation can be
   challenging. In higher dimensions (e.g. 10), this kind of visual interpretation
   becomes difficult and is rarely attempted.


## Factual Question 3

**Explain, in overview, how PCA works. Use Figure 5.4 on page 224 in your explanation.**

Covered in Section 5.3, *"Principalkomponentanalys (PCA)"* (p. 223-225).

**The core idea.** To perform a dimensionality reduction, data points can be
*projected* down onto a lower-dimensional hyperplane (a hyperplane is the
generalization, to any number of dimensions, of a line in 2D or a plane in 3D - see
the book's info box on p. 223). The key question is *how* to choose that hyperplane.
The book's answer: *"a reasonable way to do this is to choose the hyperplane that
retains as much information from the original dataset as possible. To do this, PCA
chooses the hyperplane that retains the highest share of variance"* (p. 223).

**Walking through Figure 5.4 (p. 224).** The left-hand panel of Figure 5.4 shows a
2-dimensional dataset (axes $x_1$ and $x_2$) together with three candidate lines that a
1-dimensional projection could be made onto: a solid line, a dashed line, and a dotted
line. The right-hand panel shows the result of projecting every point onto each of
those three lines in turn. Comparing the three right-hand strips, the points projected
onto the **solid line** are the most spread out - i.e. that projection retains the most
variance - followed by the dashed line, and finally the dotted line, whose projected
points are the most tightly bunched together (retaining the least variance). Since PCA
always wants to retain as much variance as possible, **the data would be projected onto
the solid line**.
**Finding the principal components.** Generalizing this idea, PCA works as follows:

1. It first identifies the axis that retains the **largest** share of variance in the
   original data. This axis does not need to align with any of the original axes - it
   points in whichever direction the data happens to spread out the most. In Figure
   5.4 this is the axis labelled $C_1$ (which corresponds to the solid line just
   discussed).
2. It then identifies a **second** axis that retains the next-largest share of
   variance, under one constraint: it must be **orthogonal** (perpendicular) to the
   first axis. This is $C_2$ in Figure 5.4. The book explicitly notes that although the
   dashed line by itself retains more variance than $C_2$ does, $C_2$ cannot be the
   dashed line, precisely because the dashed line is *not* orthogonal to $C_1$ - and
   orthogonality to all previous axes is a hard requirement (p. 223-224).
3. This continues - each new axis retaining the next-largest remaining share of
   variance, always orthogonal to every axis found so far - until there are as many new    axes as there were dimensions in the original data. **Each of these axes is called a principal component.**
